```python
import pandas as pd
import numpy as np
import random

# 1. CREATE THE DUMMY DATA FOR CAMP LEMONNIER
np.random.seed(42) # This keeps our random numbers the same every time
vehicle_types = ['NTV', 'MHE', 'SANY', 'JLG']
statuses = ['Operational', 'In Maintenance', 'Pending Parts']

data = {
    'Vehicle_ID': ['CL-V-' + str(i).zfill(3) for i in range(1, 51)],
    'Type': np.random.choice(vehicle_types, 50),
    'Mileage': np.random.randint(1000, 50000, 50),
    'Daily_Lease_Cost_USD': np.random.uniform(80, 450, 50).round(2),
    'Maintenance_Status': np.random.choice(statuses, 50, p=[0.7, 0.2, 0.1]),
    'Fuel_Consumed_Gal': np.random.uniform(5, 60, 50).round(1)
}

# Turn the data into a Pandas "DataFrame" (an invisible Excel sheet)
df = pd.DataFrame(data)

# 2. ANALYZE THE DATA (The Business Logic)
# We want to know the average cost and count of each vehicle type
summary_df = df.groupby('Type').agg(
    Total_Vehicles=('Vehicle_ID', 'count'),
    Avg_Daily_Cost_USD=('Daily_Lease_Cost_USD', 'mean'),
    Total_Fuel_Gal=('Fuel_Consumed_Gal', 'sum')
).round(2).sort_values(by='Avg_Daily_Cost_USD', ascending=False)

# Show the summary on the screen
print("--- CAMP LEMONNIER: LEASED FLEET COST SUMMARY ---")
display(summary_df)

# 3. EXPORT THE RESULTS
# Save both the raw data and our summary as CSV files
df.to_csv('Camp_Lemonnier_Raw_Data.csv', index=False)
summary_df.to_csv('Camp_Lemonnier_Cost_Summary.csv')

print("\n✅ SUCCESS: Data analyzed and exported!")
```

```python
import pandas as pd

# 1. Ingest the clean summary data from Project 1
df_emissions = pd.read_csv('Camp_Lemonnier_Cost_Summary.csv')

# 2. Define the standard EPA emissions multiplier 
# (1 gallon of diesel = 10.18 kg of CO2)
diesel_co2_factor = 10.18 

# 3. Create the automated Carbon Footprint column
# Note: Check your CSV to ensure the column is named 'Total_Fuel' or 'Total_Fuel_Gallons'. Adjust the name below if needed!
df_emissions['Total_CO2_kg'] = df_emissions['Total_Fuel_Gal'] * diesel_co2_factor

# 4. Sort the data so the worst polluter is at the very top
df_emissions = df_emissions.sort_values(by='Total_CO2_kg', ascending=False)

# 5. Export the official ESG Report for the manager
df_emissions.to_csv('Camp_Lemonnier_ESG_Report.csv', index=False)

# Display the final table in VS Code
print("CAMP LEMONNIER: CARBON EMISSIONS TRACKER")
display(df_emissions[['Type', 'Total_Fuel_Gal', 'Total_CO2_kg']])
```
